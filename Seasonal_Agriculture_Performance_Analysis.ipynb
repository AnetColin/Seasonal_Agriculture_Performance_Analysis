# Problem Statment

Agricultural activities are influenced by seasonal variations in environmental conditions, 
farming practices, resource availability and market conditions. As a result, agricultural 
performance may differ from one season to another. 
However, raw agricultural data does not clearly explain how agricultural performance changes 
across seasons or what patterns can be observed in different seasonal conditions. 
The problem is to analyze the given agricultural dataset and investigate seasonal differences 
in agricultural performance by identifying meaningful patterns, trends, relationships and 
variations within the available data.

```python
season_yield_stats = (
    df_clean
    .groupby("Season")["Yield_Tonnes_Ha"]
    .agg(
        Mean="mean",
        Median="median",
        Std_Dev="std",
        Minimum="min",
        Maximum="max",
        Records="count"
    )
    .round(2)
)

print("Yield Statistics by Season:")
display(season_yield_stats)
```

```python
from scipy.stats import f_oneway

kharif_yield = df_clean[
    df_clean["Season"] == "Kharif"
]["Yield_Tonnes_Ha"]

rabi_yield = df_clean[
    df_clean["Season"] == "Rabi"
]["Yield_Tonnes_Ha"]

zaid_yield = df_clean[
    df_clean["Season"] == "Zaid"
]["Yield_Tonnes_Ha"]

anova_result = f_oneway(
    kharif_yield,
    rabi_yield,
    zaid_yield
)

print("One-Way ANOVA — Seasonal Yield")
print("--------------------------------")
print("F-statistic:", round(anova_result.statistic, 4))
print("p-value:", anova_result.pvalue)
```

A one-way ANOVA was conducted to examine whether average agricultural yield differed significantly across Kharif, Rabi, and Zaid seasons. The test produced an F-statistic of 1.5439 and a p-value of 0.2137. Since the p-value was greater than 0.05, the null hypothesis was not rejected. Therefore, the analysis did not provide sufficient statistical evidence of a significant difference in mean yield among the three seasons. Although Kharif recorded the highest descriptive average yield, the substantial within-season variability means that the observed seasonal differences were not statistically significant.

```python
kharif_profit = df_clean[
    df_clean["Season"] == "Kharif"
]["Profit_INR"]

rabi_profit = df_clean[
    df_clean["Season"] == "Rabi"
]["Profit_INR"]

zaid_profit = df_clean[
    df_clean["Season"] == "Zaid"
]["Profit_INR"]

profit_anova = f_oneway(
    kharif_profit,
    rabi_profit,
    zaid_profit
)

print("One-Way ANOVA — Seasonal Profit")
print("--------------------------------")
print("F-statistic:", round(profit_anova.statistic, 4))
print("p-value:", profit_anova.pvalue)
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

tukey_profit = pairwise_tukeyhsd(
    endog=df_clean["Profit_INR"],
    groups=df_clean["Season"],
    alpha=0.05
)

print("Tukey HSD Post-Hoc Test — Seasonal Profit")
print("------------------------------------------")
print(tukey_profit)
```

```python
from scipy.stats import f_oneway

drip_yield = df_clean[
    df_clean["Irrigation_Method"] == "Drip"
]["Yield_Tonnes_Ha"]

flood_yield = df_clean[
    df_clean["Irrigation_Method"] == "Flood"
]["Yield_Tonnes_Ha"]

rainfed_yield = df_clean[
    df_clean["Irrigation_Method"] == "Rainfed"
]["Yield_Tonnes_Ha"]

sprinkler_yield = df_clean[
    df_clean["Irrigation_Method"] == "Sprinkler"
]["Yield_Tonnes_Ha"]

irrigation_yield_anova = f_oneway(
    drip_yield,
    flood_yield,
    rainfed_yield,
    sprinkler_yield
)

print("One-Way ANOVA — Irrigation Method and Yield")
print("--------------------------------------------")
print("F-statistic:", round(irrigation_yield_anova.statistic, 4))
print("p-value:", irrigation_yield_anova.pvalue)
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

tukey_irrigation_yield = pairwise_tukeyhsd(
    endog=df_clean["Yield_Tonnes_Ha"],
    groups=df_clean["Irrigation_Method"],
    alpha=0.05
)

print("Tukey HSD Post-Hoc Test — Irrigation Method and Yield")
print("------------------------------------------------------")
print(tukey_irrigation_yield)
```

### Note
Tukey's HSD post-hoc test revealed that the significant differences in yield occurred between Drip and Flood irrigation (p = 0.0148) and between Drip and Rainfed irrigation (p = 0.0058). No statistically significant difference was observed between Drip and Sprinkler or among the remaining irrigation-method pairs. This indicates that the overall variation in yield across irrigation methods is primarily associated with the higher observed yield under Drip irrigation compared with Flood and Rainfed
